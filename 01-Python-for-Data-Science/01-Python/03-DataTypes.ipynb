##### Data Types

1.  A Data Type helps in determining the type of the value that the variable is holding.

2.  It is determined automatically once the value is assigned to the variable.

3.  We can determine the type of the variable using type() function.

```python
name = 'Parag'
age  =  31
salary = 53874.23
is_employed = True

##### print the variables
print(name)
print(age)
print(salary)
print(is_employed)

##### print the data type of the variable
print(type(name))
print(type(age))
print(type(salary))
print(type(is_employed))
```

##### Types of data type

1.  Integer --------->  It is used to store only whole numbers. It has no decimal part.

```python
age = 31

print(age)

print(type(age))
```

2.  Floating ------->   It is used to store the decimal part or the fractional part.

```python
salary = 53874.23

print(salary)

print(type(salary))
```

3.  String -------->    It is used to store the text enclosed within single or double quotes.

```python
name = 'Parag'

print(name)

print(type(name))
```

4.  Boolean ------->    It is used to store either True or False

```python
x  = True

print(x)

print(type(x))
```

5. Complex -------> It has a real and imaginary part.

```python
x = 2+3j

print(x)

print(type(x))
```

6. NoneType -------> It determines the absence of the variable

```python
x = None

print(x)

print(type(x))
```

1.   Python provides the built-in type() function to identify the type of a value.

```python
age = 25
price = 99.50
name = "Parag"
is_active = True
result = None

print(type(age))
print(type(price))
print(type(name))
print(type(is_active))
print(type(result))
```

13. Given data = [10, 20.5, "30", True, None], write a program to count how many values belong to each exact built-in type present in the collection.

```python
data = [10, 20.5, "30", True, None]

freq = {}

for x in data:
    if(x in freq):
        freq[x] = freq[x] + 1
    else:
        freq[x] = 1
print(freq)
```

18. Write a program that accepts a list of values and produces a dictionary where each key is a data type name and each value is the count of items belonging to that exact type

```python
values = [10, 20.5, "30", True, None, 40, "Hello", False]

freq = {}

for x in values:
    if(type(x) in freq):
        freq[type(x)] = freq[type(x)] + 1
    else:
        freq[type(x)] = 1
print(freq)
```

20. Write a program that takes a collection of values and groups them by their exact built-in data type while preserving the original values and their order within each group

```python
values = [10, 20.5, "30", True, None, 40, "Hello", False, 15.5]

freq = {}

for x in values:
    if(type(x) in freq):
        freq[type(x)] = freq[type(x)] + 1
    else:
        freq[type(x)] = 1
print(freq)
```
