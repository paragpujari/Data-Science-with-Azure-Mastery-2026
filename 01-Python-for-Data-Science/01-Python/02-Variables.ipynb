var_name = value

```python
##### Defining the Variable

name        =   'Parag'
age         =     31
city        =   'Noida'
state       =   'Uttar Pradesh'
country     =   'India'
salary      =   53874
gender      =   'Male'
is_employed =   True

##### Print all the variables
print(f"Name    of the employee is {name}")
print(f"Age     of the employee is {age}")
print(f"City    of the employee is {city}")
print(f"State   of the employee is {state}")
print(f"Country of the employee is {country}")
print(f"Salary  of the employee is {salary}")
print(f"Gender  of the employee is {gender}")
print(f"Employability of the employee is {is_employed}")
```

##### Creating the Variable

```python
var = 10
print(var)
```

##### Reassignment

It refers to the process of changing the value of the variable to a new value.

```python
var = 10
print(var)

var = 25
print(var)

var = 60
print(var)

var = 75
print(var)
```

##### OBSERVATIONS:

1.  Here the variable var gets reassigned with different values at different instances of time.

```python
x = 10
print(x)
print(type(x))

x = 90.87
print(x)
print(type(x))

x = (2+3j)
print(x)
print(type(x))
```

##### OBSERVATIONS:

1.  The variable x has four different values defined during different runtime execution of the program.

2.  As four different values get assigned to the same variable during different runtime execution of the program, so four different datatypes gets determined automatically.

##### Rules for defining the Variable

1.  A Variable name must have letters, numbers or an underscore.

```python
first_number1 = 10
print(first_number1)
```

2. It must start with a letter or an underscore

```python
x = 10
print(x)

_x = 20
print(_x)
```

3. It must follow a snake case pattern

```python
first_name = 'Parag'
print(first_name)
```

4. Variables are case sensitive

```python
a = 10
print(a)

A = 15
print(A)
```

5. It should not begin with a number

```python
# 1age = 31  ##### Invalid : Variable should not begin with a number
```

6.  It should not have spaces or hyphen in between

```python
#first name = 'Parag'     ##### Invalid : Variables should not have spaces in between them
#print(first name)

#last-name = 'Pujari'     ##### Invalid : Variables should not have hyphen in between them
#print(last-name)
```

7.  A Variable should not have any special symbols like $,!,@ etc

```python
#$name = 'Parag'            ##### Invalid : Variables should not have $ in  them
#print($name)

#!name = 'Anish'             ##### Invalid : Variables should not have ! in  them
#print(!name)

#@name = 'Ayush'             ##### Invalid : Variables should not have @ in  them
#print(@name)
```

```python
x = 10
y = 20

print(f'Value of x before swapping is {x}')
print(f'Value of y before swapping is {y}')

##### Swapping of two variables
x,y = y,x

print(f'Value of x after swapping is {x}')
print(f'Value of y after swapping is {y}')
```

40. You have the following variables:

original_price = 1000

discount = 200

tax = 80

Calculate the final price using only variables and arithmetic.

The logic is:

Price after discount = original price - discount

Final price = price after discount + tax

```python
original_price = 1000

discount = 200

tax = 80

##### Calculate the price after discount
price_after_discount = original_price - discount
print(f'Price after discount is {price_after_discount}')

##### Calculate the final price
final_price = price_after_discount + tax
print(f'Final Price is {final_price}')
```
